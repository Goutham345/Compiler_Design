```python
%%writefile octal.l
%{
#include <stdio.h>
#include <stdlib.h>
%}

%%
[0-7]+ {
    int decimal = 0;
    int i;

    for (i = 0; yytext[i] != '\0'; i++) {
        decimal = decimal * 8 + (yytext[i] - '0');
    }

    printf("Octal Number  : %s\n", yytext);
    printf("Decimal Number: %d\n", decimal);
}

[ \t\n]+ ;

. {
    printf("Invalid octal number!\n");
}
%%

int yywrap()
{
    return 1;
}

int main()
{
    printf("Enter an octal number: ");
    yylex();
    return 0;
}
```

```python
%%writefile analyzer.l
%{
#include <stdio.h>

int vowels = 0;
int consonants = 0;
int characters = 0;
int words = 0;
int lines = 0;
%}

%%
[aeiouAEIOU] {
    vowels++;
    characters++;
}

[a-zA-Z]+ {
    int i;

    words++;

    for (i = 0; i < yyleng; i++) {
        if (yytext[i]=='a' || yytext[i]=='e' || yytext[i]=='i' ||
            yytext[i]=='o' || yytext[i]=='u' ||
            yytext[i]=='A' || yytext[i]=='E' || yytext[i]=='I' ||
            yytext[i]=='O' || yytext[i]=='U')
            vowels++;
        else
            consonants++;
    }

    characters += yyleng;
}

[0-9]+ {
    words++;
    characters += yyleng;
}

[ \t]+ {
    characters += yyleng;
}

\n {
    characters++;
    lines++;
}

. {
    characters++;
}
%%

int yywrap()
{
    return 1;
}

int main()
{
    FILE *file;

    file = fopen("input.txt", "r");

    if (file == NULL) {
        printf("Error: Cannot open input.txt\n");
        return 1;
    }

    yyin = file;
    yylex();

    fclose(file);

    printf("\n----- Text File Analysis -----\n");
    printf("Vowels      : %d\n", vowels);
    printf("Consonants  : %d\n", consonants);
    printf("Characters  : %d\n", characters);
    printf("Words       : %d\n", words);
    printf("Lines       : %d\n", lines);

    return 0;
}
```
